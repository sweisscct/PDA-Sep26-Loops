# Loops
There are two types of loops in Python: `for` loops and `while` loops. We use them when we want to repeat some code.

## `for` Loops
These are used when we want to repeat something a predetermined number of times

eg we have 100 row of data and we want to do something to each one.

## `while` loops
These are used when we don't know how many times (if any) we are going to repeat some code.

```python
# Programming languages tend to start counting at 0
count = 0
# range(10) means we are going to assign 0-9 to the variable num

# num = 0
# print(num)
# count += 1

# num = 1
# print(num)
# count += 1

# num = 2
# print(num)
# count += 1

# ...

# num = 9
# print(num)
# count += 1

# print(f"The count is: {count}")

for num in range(10):
    print(num)
    count += 1
print(f"The count is: {count}")
```

## `range()`
- range(n) means all the numbers from 0 to n-1
- range(a, n) means all the numbers from a to n-1
- range(a, n, x) means all the numbers from a to n-1, going up in increments of x

```python
count = 0
for num in range(5, 10):
    print(num)
    count += 1
print(f"The count is: {count}")
```

```python
count = 0
for num in range(5, 106, 10):
    print(num)
    count += 1
print(f"The count is: {count}")
```

```python
count = 0
for num in range(105, 5, -10):
    print(num)
    count += 1
print(f"The count is: {count}")
```

## `while` loops

```python
# while condition:
#     ... code

# as long as the condition evaluates to True, 
# the loop will keep resetting to the top when it reaches the end

# Note: this means we may get stuck in an infinite loop!!
```

```python
num = 50
while num < 10:
    print(num)
    num += 1
print(f"The end result is: {num}")
```

## Practice game!
- Guessing game
- The game generates a random number
- The user has to keep guessing until they guess the number
- The user gets told if they guessed too high or too low

```python
import random
low_num = 1
high_num = 100
correct_num = random.randint(low_num, high_num)
print(f"The correct number is: {correct_num}")

guess = int(input("What is your next guess?"))
while guess != correct_num:
    if guess > correct_num:
        print("Sorry, that's too high!")
    else:
        print("That's too low, please try again.")
    guess = int(input("What is your next guess?"))

print(f"Congratulations!!")
```
